```python
# 일반 for문
# 빠른 for문

data1 = "apple" + "banana"

# 순차적으로 순회
# 객체의 개수만큼 반복
for i in data1:
    print(i)
```

```python
for i in [10, 20, 30, 40]:
    print(i)
```

## 컴프리헨션

```python
[i * 2 for i in [10, 20, 30]]
```

```python
{i * 2 for i in [10, 20, 30]}
```

```python
name = "장보고", "홍길동", "이순신"
```

```python
for i in name:
    print(i)
```

```python
# enumerate: 인덱스를 만드는 함수
enumerate(name)
```

```python
# 일반 for문(range)
# range(10)
# range(10, 20)
# range(20, 30, 2)

for i in range(10):
    print(i)
```

```python
# 1~50까지 홀수만 출력
for i in range(1, 51, 2):
        print(i)
```

```python
# 1~50까지 홀수만 리스트에 담아서 출력
[i for i in range(1, 51, 2)]
```

```python
# 1~50까지 반복하여
# 세트 컴프리헨션에 0,1,2,3,4,5만 담아서 출력

{i % 6 for i in range(51)}
```

## 딕셔너리 빠른 for문

```python
foods = {
    "a": "카레",
    "b": "라면",
    "c": "불고기",
    "d": "떡볶이",
}
```

```python
for i in foods:
    print(i)
```

```python
foods.values()
```

```python
for key, food in foods.items():
    print(f"{key} {food}")
```
