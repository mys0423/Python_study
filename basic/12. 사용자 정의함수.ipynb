```python
def order_info(*args, name, address, phone):
    total = sum(args)
    print(f"{name}님의 주소: {address}, 연락처: {phone}")
    print(f"{name}님의 총 구매 가격: {total}")
```

```python
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림
# 사용자의 정보가 입력되면 같이 출력하기
def get_average(*args, **keywards):
    total = sum(args)
    average = round(total / len(args), 1)
    print("이름: " + keywards["name"])
    print(f"평균: {average}")
    
```

```python
get_average(100, 20, 30, name="홍길동")
```

```python
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림
# 사용자의 정보가 입력되면 같이 출력하기
def get_average(**kwards):
    kor, eng, math = kwards.get("kor", 0), kwards.get("eng", 0), kwards.get("math", 0)
    total = kor + eng + math

    average = total / 3
    
    if "round" in kwards:
        average = round(average, kwards["round"])
    
    if "name" in kwards:
        name = kwards["name"]
        return f"{name}님의 평균은: {average}"

    return f"평균은: {average}"
```

```python
user_info = {
    "kor": 100,
    "eng": 80,
    "math": 22,
    "name": "홍길동",
    "round": 5
}
```

```python
get_average(**user_info)
```
