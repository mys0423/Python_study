## 커스텀 예외(Custom Exception)

```python
# Exception: 모든 예외의 부모 클래스
class BadWordException(Exception):
    def __init__(self, message):
        super().__init__(message)
```

```python
# 사용자가 "바보"라고 입력을 하면 커스텀 예외 발생!
try:
    expression = input("채팅을 입력하세요")

    if "바보" in expression:
        raise BadWordException("바보라고 하지마!🫵")

except BadWordException as e:
    print(e)
```

```python
# 국어, 영어, 수학 점수를 받아서 총점과 평균을 구해주는 프로그램
# 0 ~ 100점 까지만 입력할 수 있습니다.
# 이름을 입력할 수도 있습니다.
# 학번을 입력할 수도 있습니다.
# 평균의 소숫점 자리를 입력받아 처리할 수도 있습니다.
# 예외가 발생한다면 try ~ except 문법으로 해결하세요.

class Average:
    def __init__(self, kor, eng, math, name=None, student_id=None):
        self.name = name
        self.student_id = student_id

        try:
            self.kor = int(kor)
            self.eng = int(eng)
            self.math = int(math)

            for score in [self.kor, self.eng, self.math]:
                if not (0 <= score <= 100):
                    raise ValueError("점수는 0~100 사이여야 합니다.")

        except ValueError as e:
            print(f"입력 오류: {e}")
            self.kor = self.eng = self.math = 0

        self.total = self.kor + self.eng + self.math
        self.avg = round(self.total / 3)

    def result(self):
        if self.name:
            print(f"이름: {self.name}")
        if self.student_id:
            print(f"학번: {self.student_id}")
        print(f"국어: {self.kor}, 영어: {self.eng}, 수학: {self.math}")
        print(f"총점: {self.total}")
        print(f"평균: {self.avg}")


s = Average(90, 85, 78, name="홍길동", student_id="20240001")
s.result()

            
        
```

```python
# 실습
# 국어, 영어, 수학 점수를 받아서 총점과 평균을 구해주는 프로그램
# 0 ~ 100점 까지만 입력할 수 있습니다.
# 이름을 입력할 수도 있습니다.
# 학번을 입력할 수도 있습니다.
# 평균의 소숫점 자리를 입력받아 처리할 수도 있습니다.
# 예외가 발생한다면 try ~ except 문법으로 해결하세요.

class Student:
    def __init__(self, student_id, name, kor, eng, math):
        self.student_id = student_id
        self.name = name
        self.kor = kor
        self.eng = eng
        self.math = math
        self.total = kor + eng + math
        self.average = self.total / 3
    
    def print_score(self, **kwards):
        is_name_print = kwards.get("name", False)
        round_number = kwards.get("round", 0)

        name = f"{self.name}님의 " if is_name_print else ""
        average = round(self.average, round_number) if round_number else int(self.average)
        message = f"{name}총 점수는: {self.total}점, 평균은: {average}점"
        print(message)
```

```python
hong = Student(1, "홍길동", 40, 50, 67)

hong.print_score(name=True, round=3)
```

```python
try:
    kor = int(input("국어 점수를 입력하세요.\nex) 75"))
    math = int(input("수학 점수를 입력하세요.\nex) 75"))
    eng = int(input("영어 점수를 입력하세요.\nex) 75"))
    
    if not (0 <= kor <= 100) or not (0 <= math <= 100) or not (0 <= eng <= 100):
        raise ScoreOutOfRangeException("0 ~ 100까지의 점수를 입력하세요.")

    hong = Student(1, "홍길동", kor, math, eng)

    is_name = eval(input("이름을 출력할까요?\nex)True\n"))
    is_round = int(input("소수점 자리 출력값을 입력하세요.\nex)0"))
    payload = {}
    
    if is_name:
        payload.update({"name": is_name})

    payload.update({"round": is_round})

    hong.print_score(**payload)

except ScoreOutOfRangeException as e:
    print(e)

except ValueError:
    print("타입이 잘못 되었습니다.")
```
