```python
# 완전 객체 지향형 언어
# market
# 판매
class Market:
    def __init__(self, market_name, product_name, product_price):
        self.market_name = market_name
        self.product_name = product_name
        self.product_price = product_price

    def sell(self, customer):
        customer.money -= self.product_price * (100 - customer.discount) / 100

# customer
# 할인율
class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
cu = Market("CU", "도시락", 10000)
customer = Customer("이규혁", 50000, 10)

cu.sell(customer)
print(customer.money)
```

```python
# Atm, User 객체
# deposit

# Atm.deposit(User, 10000)
# 유저의 돈은 10000 차감,
# 은행에 돈은 10000 입금

# 은행이 가진돈과 유저의 돈 출력

class Atm:
    def __init__(self, bank_name, bank_money):
        self.bank_name = bank_name
        self.bank_money = bank_money

    def deposit(self, user, amount):
        user.money -= amount
        self.bank_money += amount

        print(f"은행 돈: {self.bank_money}")
        print(f"유저 돈: {user.money}")
    
class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money
```

```python
user = User("홍길동", 50000)
atm = Atm("국민은행", 100000)

atm.deposit(user, 10000)
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

class Store:
    def __init__(self, name):
        self.name = name
        self.products = {}
        self.income = 0

    def register(self, product):
        if product.name in self.products:
            print(f"{product.name}은 이미 등록된 상품입니다.")
            return
        self.products[product.name] = product
        print(f"{product.name} 등록 완료.")

    def sell(self, product_name, quantity, user):
        if product_name not in self.products:
            print("존재하지 않는 상품입니다.")
            return

        product = self.products[product_name]

        if product.stock < quantity:
            print("재고가 부족합니다.")
            return

        total = product.price * quantity

        if user.money < total:
            print("잔액이 부족합니다.")
            return

        product.stock -= quantity
        user.money -= total
        self.income += total
        print(f"{user.name}이 {product_name}을 {quantity}개 구매. 잔액: {user.money}")

    def modify(self, product_name, name=None, price=None, stock=None):
        if product_name not in self.products:
            print("존재하지 않는 상품입니다.")
            return

        product = self.products[product_name]

        if name:
            self.products[name] = self.products.pop(product_name)
            product.name = name
        if price is not None:
            product.price = price
        if stock is not None:
            product.stock = stock

        print(f"{product_name} 수정 완료.")

    def remove(self, product_name):
        if product_name not in self.products:
            print("존재하지 않는 상품입니다.")
            return
        del self.products[product_name]
        print(f"{product_name} 삭제 완료.")

class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money
```

```python
# 클래스 메서드
class Korean:
    # 클래스 변수
    country = "한국"

    # 클래스 메서드
    # 인스턴스x, 클래스변수o
    @classmethod # 데코레이터
    def trip(cls, country):
        if cls.country == country:
            print("국내여행")
        else:
            print("해외여행")

    # 정적 메서드
    @staticmethod
    def slogan():
        print("Imagine your korea")
```

```python
hong = Korean()

hong.trip("한국")
hong.slogan()

Korean.trip("미국")
Korean.slogan()
```

## 상속(inheritance)

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age
```

```python
hong = Person("홍길동")
jang = Student("장보고", 20)

# 부모 -> 타입도 같이 전달!
# 모든 자식은 부모 타입이다.
# 부모는 자식 타입이 아니다.
print(isinstance(hong, Person))
print(isinstance(hong, Student))
print(isinstance(jang, Person))
print(isinstance(jang, Student))
```

```python
class Animal:
    def __init__(self, name, feed, species):
        self.name = name
        self.feed = feed
        self.species = species

    def feeding(self):
        print(f"{self.name}은 먹이를 먹습니다.")
        
class Dog(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def act(self):
        print(f"{self.name}은 산택을 합니다")

class Cat(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def sleep(self):
        print(f"{self.name}은 잠을 자주 잡니다.")
        
```

```python
dog = Dog("누렁이", "잡식", "강아지")

dog.act()

cat = Cat("야옹이", "츄르", "고양이")
cat.sleep()
```

```python
# 사람(부모), 학생(자식)
# 사람 - 이름, 나이, 키, 몸무게
# 학생 - 학생, 핸드폰

# 사람은 인사하는 메서드
# 학생은 공부하는 메서드

class Person:
    def __init__(self, name, age, height, weight):
        self.name = name
        self.age = age
        self.height = height
        self.weight = weight

    def hi(self):
        print("안녕하세용")

class Student(Person):
    def __init__(self, name, age, height, weight, phone):
        super().__init__(name, age, height, weight)

    def study(self):
        print("공부 하는중....")
```

```python
jang = Person("장보고", 20, 180, 100)
hong = Student("홍길동", 25, 180, 80, "010-2222-3333")
```

```python
jang.hi()
hong.hi()
hong.study()
```

## 다형성

```python
class Car:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def start_engine(self):
        print(f"{self.brand}는 열쇠로 시동을 켭니다")

    def stop_engine(self):
        print(f"{self.brand}는 열쇠로 시동을 끕니다")
        
class SuperCar(Car):
    def start_engine(self):
        print(f"{self.brand}는 음성으로 시동을 켭니다")

    def stop_engine(self):
        print(f"{self.brand}는 음성으로 시동을 끕니다")
    
```

```python
matiz = Car("마티즈", 10000000)
matiz.start_engine()
matiz.stop_engine()
```

```python
lamborghini = SuperCar("람보르기니", 1000000000)
lamborghini.start_engine()
lamborghini.stop_engine()
```

```python
# 휴대폰 필드
# 브랜드(brand), 가격(price)
# 2G 전화하기
# 3G 전화하기, 인터넷
# 4G 전화하기, 인터넷, 영상통화

class Phone:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

class Phone2G(Phone):
    def call(self):
        print("전화 연결중...")

class Phone3G(Phone2G):
    def internet(self):
        print("인터넷 연결중...")

class Phone4G(Phone3G):
    def face_call(self):
        print("영상통화 연결중...")
```

```python
galaxy1 = Phone2G("삼성", 10000)
galaxy2 = Phone3G("삼성", 20000)
galaxy3 = Phone4G("삼성", 30000)

galaxy1.call()

galaxy2.call()
galaxy2.internet()

galaxy3.call()
galaxy3.internet()
galaxy3.face_call()
```

```python
# 휴대폰 필드
# 브랜드(brand), 가격(price)
# 2G 전화하기
# 3G 전화하기, 인터넷
# 4G 전화하기, 인터넷, 영상통화

class Phone2G:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def call(self):
        print(f"{self.brand}은 일반 전화")

class Phone3G(Phone2G):
    def __init__(self, brand, price):
        super().__init__(brand, price)

    def call(self):
        print(f"{self.brand}은 영상 통화도 가능")

    def browse(self):
        print("인터넷 느리게 사용 가능")

class Phone4G(Phone3G):
    def __init__(self, brand, price):
        super().__init__(brand, price)

    def browse(self):
        print("인터넷 사용 가능(속도 개선)")

    def stream_vedio(self):
        print("영상 통화 가능")
```
