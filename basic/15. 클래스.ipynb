```python
# 이름(name), 나이(age), 운동종목(sports)
# 플레이어들을 소개하는 메서드

# 운동선수 3명의 객체화 하고
# 3명을 각각 소개하는 메서드를 사용해서 정보들을 출력하세요.
# ex) OOO선수의 나이 OO살 운동종목 OOOO
class Player:
    def __init__(self, name, age, sports):
        self.name = name
        self.age = age
        self.sports = sports
    
    def introduce(self):
        print(f"{self.name} 선수의 나이 {self.age}살 운동종목 {self.sports}")
```

```python
# 완전 객체 지향형 언어
# market
# 판매
class Market:
    def __init__(self, market_name, product_name, product_price):
        self.market_name = market_name
        self.product_name = product_name
        self.product_price = product_price

    def sell(self, customer):
        customer.money -= self.product_price * (100 - customer.discount) / 100

# customer
# 할인율
class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
cu = Market("CU", "도시락", 10000)
customer = Customer("이규혁", 50000, 10)

cu.sell(customer)
print(customer.money)
```

```python
# Atm, User 객체
# deposit

# Atm.deposit(User, 10000)
# 유저의 돈은 10000 차감,
# 은행에 돈은 10000 입금

# 은행이 가진돈과 유저의 돈 출력

class Atm:
    def __init__(self, bank_name, bank_money):
        self.bank_name = bank_name
        self.bank_money = bank_money

    def deposit(self, user, amount):
        user.money -= amount
        self.bank_money += amount

        print(f"은행 돈: {self.bank_money}")
        print(f"유저 돈: {user.money}")
    
class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money
```

```python
user = User("홍길동", 50000)
atm = Atm("국민은행", 100000)

atm.deposit(user, 10000)
```

```python

            # 상품 등록
    def register(self, product_name, price, stock):
        # 기존의 상품이 있다면 추가하지 않는다.(유효성 검사)
        if product_name in self.products:
            print(f"❌ 이미 '{product_name}' 상품이 등록되어 있습니다.")
        # 기존의 상품이 없다면 추가
        else:
            self.products[product_name] = {"price": price, "stock": stock}
            print(f"✅ '{product_name}' 등록 완료! (가격: {price}원, 재고: {stock}개)")

    # 상품 판매
    def sell(self, user, product_name, amount):
        # 1. 상품이 상점에 있는지 확인
        if product_name not in self.products:
            print(f"❌ '{product_name}' 상품은 판매하지 않습니다.")
            return

        product = self.products[product_name]
        total_price = product["price"] * amount

        # 2. 재고 확인
        if product["stock"] < amount:
            print(f"❌ 재고가 부족합니다. (현재 재고: {product['stock']}개)")
            return

        # 3. 구매자의 잔액 확인
        if user.money < total_price:
            print(f"❌ 잔액이 부족합니다. (필요 금액: {total_price}원, 소지 금액: {user.money}원)")
            return

        # --- 판매 처리 로직 ---
        # 구매자의 money 감소
        user.money -= total_price
        # 수익 store 저장
        self.income += total_price
        # 상품의 stock 구매 수량만큼 제거
        product["stock"] -= amount
        
        print(f"🛒 [판매 완료] {user.name}님이 '{product_name}' {amount}개를 구매했습니다. (결제: {total_price}원)")

    # 상품 수정 (이름, 가격, 재고 수정)
    def modify(self, old_name, new_name, new_price, new_stock):
        if old_name not in self.products:
            print(f"❌ 수정할 '{old_name}' 상품이 존재하지 않습니다.")
            return
        
        # 기존 상품 정보를 삭제하고, 새로운 이름과 정보로 다시 등록합니다.
        # (이름을 변경하지 않는 경우 old_name == new_name)
        if old_name != new_name:
            del self.products[old_name]
            
        self.products[new_name] = {"price": new_price, "stock": new_stock}
        print(f"🔄 상품 수정 완료: '{new_name}' (가격: {new_price}원, 재고: {new_stock}개)")

    # 상품 삭제
    def remove(self, product_name):
        if product_name in self.products:
            del self.products[product_name]
            print(f"🗑️ '{product_name}' 상품이 삭제되었습니다.")
        else:
            print(f"❌ 삭제할 '{product_name}' 상품이 존재하지 않습니다.")


class User:
    def __init__(self, name, money):
        # 변수에 값을 할당하도록 수정했습니다.
        self.name = name
        self.money = money
```
